````python
#if/else
if helado.lower()=="si" : print ("Te lo compro")
elif helado.lower()=="sí" : print ("Te lo compro")
elif helado.lower()=="sipi" : print ("Te lo compro")
else:print("No te lo ocmpro")
````

##Match/case
La estructura Match evalua una varaibale frente a varios bloques case.El comodin case _ funciona como ninguna de las anteriores.

Sintaxis
```
match (condicion):
 case v1: aqui se ejecuta si condicion==v1
 case v2: aqui se ejecuta si condicion==v2
 case v3: aqui se ejecuta si condicion==v3
 case_:ninguna de las anteriores
```
