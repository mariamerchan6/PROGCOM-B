```python
print("==============================================")
print("  EVALUADOR DE SALUD FINANCIERA")
print("  ¿Puedes emprender sin quebrarte?")
print("==============================================")

nombre = input("Tu nombre: ")
ingresos = float(input("Cuanto ganas al mes? $: "))


entrada_gastos = input("Cuanto gastas fijo al mes? (puedes poner varios con comas) $: ")
gastos_fijos = 0
if "," in entrada_gastos:
    for valor in entrada_gastos.split(","):
        gastos_fijos = gastos_fijos + float(valor)
else:
    gastos_fijos = float(entrada_gastos)

deudas = float(input("Cuanto pagas en deudas al mes? $: "))
ahorro_actual = float(input("Cuanto tienes ahorrado hoy? $: "))
costo_emprendimiento = float(input("Cuanto necesitas para empezar tu negocio? $: "))

dinero_disponible = ingresos - gastos_fijos - deudas
porcentaje_endeudamiento = (deudas / ingresos * 100) if ingresos > 0 else 0
meses_aguante = ahorro_actual / gastos_fijos if gastos_fijos > 0 else 0

print("\n--- DIAGNOSTICO ---")
print(f"Te quedan libres al mes: ${dinero_disponible:,.0f}")
print(f"Nivel de endeudamiento: {porcentaje_endeudamiento:.1f}%")
print(f"Con tus ahorros aguantas: {meses_aguante:.1f} meses sin ingresos")

if dinero_disponible <= 0:
    print("\nNO VIABLE AHORA: Gastas mas de lo que ganas.")
elif porcentaje_endeudamiento > 40:
    print("\nRIESGO ALTO: Debes mas del 40% de lo que ganas.")
elif ahorro_actual < costo_emprendimiento:
    print(f"\nTE FALTA PLATA: Necesitas ${costo_emprendimiento:,.0f} y tienes ${ahorro_actual:,.0f}")
else:
    print(f"\nVIABLE: Puedes empezar.")
    print(f"Te quedan ${ahorro_actual - costo_emprendimiento:,.0f} de colchon.")

```
