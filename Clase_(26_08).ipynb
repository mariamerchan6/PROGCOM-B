```python
agenda = []

while True:
    print("\n==== GUILD WARS - AGENDA =====")
    print("1. Ver agenda")
    print("2. Agregar actividad")
    print("3. Buscar actividad")
    print("4. Eliminar actividad")
    print("5. Mostrar cantidad de actividades")
    print("6. Ver primera actividad")
    print("7. Ver última actividad")
    print("8. Ordenar agenda")
    print("9. Salir")

    opcion = input("Elige una opción: ")

    match opcion:
        case "1":
            if len(agenda) == 0:
                print("La agenda está vacía")
            else:
                print("\nActividades en la agenda:")
                for i in range(len(agenda)):
                    print(f"{i + 1}. {agenda[i]}")

        case "2":
            actividad = input("Escribe la actividad a agregar: ")
            agenda.append(actividad)
            print(f"'{actividad}' fue agregada a la agenda")

        case "3":
            buscar = input("Escribe la actividad a buscar: ")

            if buscar in agenda:
                print(f"'{buscar}' sí está en la agenda")
            else:
                print(f"'{buscar}' no se encontró en la agenda")

        case "4":
            eliminar = input("Escribe la actividad a eliminar: ")

            if eliminar in agenda:
                agenda.remove(eliminar)
                print(f"'{eliminar}' fue eliminada")
            else:
                print(f"'{eliminar}' no existe en la agenda")

        case "5":
            print(f"Cantidad de actividades: {len(agenda)}")

        case "6":
            if len(agenda) == 0:
                print("La agenda está vacía")
            else:
                print(f"Primera actividad: {agenda[0]}")

        case "7":
            if len(agenda) == 0:
                print("La agenda está vacía")
            else:
                print(f"Última actividad: {agenda[-1]}")

        case "8":
            agenda.sort()
            print("Agenda ordenada alfabéticamente")

        case "9":
            print("Saliendo del programa... ¡Nos vemos!")
            break

        case _:
            print("Opción no válida, intenta de nuevo")
```
